# 06 — Build plot data and retained R figures


```r
suppressPackageStartupMessages({
  library(ggplot2)
  library(dplyr)
  library(tidyr)
})

SELECTED_BATCH <- ""
EXPECTED_DIMENSIONS <- 5
EXPECTED_BATCHES <- 20
TOP_FEATURES <- 12

LABEL_MAP <- c(
  FIN = "FIN/EST",
  FINLAND = "FIN/EST",
  EST = "FIN/EST",
  ESTONIA = "FIN/EST",
  RUS = "RUS",
  RUSSIA = "RUS"
)
GROUP_ORDER <- c("FIN/EST", "RUS")
GROUP_COLORS <- c("FIN/EST" = "#0072E8", "RUS" = "#F03B20")

normalize_group <- function(x) {
  key <- toupper(trimws(as.character(x)))
  out <- LABEL_MAP[key]
  out[is.na(out)] <- trimws(as.character(x))[is.na(out)]
  out
}

root <- if (dir.exists("data")) "." else ".."

read_table <- function(path) {
  if (grepl("\\.gz$", path)) read.csv(gzfile(path), check.names = FALSE) else read.csv(path, check.names = FALSE)
}

find_table <- function(folder, stem) {
  paths <- c(
    file.path(folder, paste0(stem, ".csv.gz")),
    file.path(folder, paste0(stem, ".csv"))
  )
  existing <- paths[file.exists(paths)]
  if (length(existing)) existing[1] else NA_character_
}

factor_names <- function(x) {
  grep("^factor_[0-9]+$", names(x), value = TRUE)
}

run_is_complete <- function(run, method) {
  batches <- list.dirs(run, recursive = FALSE)
  batches <- sort(batches[grepl("^batch_[0-9]+$", basename(batches))])
  if (length(batches) != EXPECTED_BATCHES) return(FALSE)

  for (folder in batches) {
    loading_path <- find_table(folder, "feature_loadings")
    score_path <- find_table(folder, "test_subject_scores")
    if (is.na(loading_path) || is.na(score_path)) return(FALSE)

    if (length(factor_names(read_table(loading_path))) != EXPECTED_DIMENSIONS) return(FALSE)

    if (method == "MEFISTO") {
      train_path <- find_table(folder, "train_sample_factors")
      test_path <- find_table(folder, "test_sample_factors")
      if (is.na(train_path) || is.na(test_path)) return(FALSE)
      if (length(factor_names(read_table(train_path))) != EXPECTED_DIMENSIONS) return(FALSE)
      if (length(factor_names(read_table(test_path))) != EXPECTED_DIMENSIONS) return(FALSE)
    }
  }
  TRUE
}

newest_five_dimension_run <- function(folder, method) {
  runs <- sort(list.dirs(folder, recursive = FALSE), decreasing = TRUE)
  for (run in runs) {
    if (run_is_complete(run, method)) return(run)
  }
  stop(paste("No complete five-dimension", method, "run found in", folder))
}

load_loadings <- function(folder) {
  path <- find_table(folder, "feature_loadings")
  if (is.na(path)) stop(paste("Missing feature_loadings in", folder))
  x <- read_table(path)
  names(x)[1] <- "feature_id"
  x
}

tempted_run <- newest_five_dimension_run(file.path(root, "data", "tempted"), "TEMPTED")

# Select the newest complete MEFISTO run that used exactly the same saved split set.
tempted_metrics <- read_table(file.path(tempted_run, "all_metrics.csv"))
if (!("split_run" %in% names(tempted_metrics)) || length(unique(tempted_metrics$split_run)) != 1) {
  stop("TEMPTED run does not contain one explicit split_run identifier")
}
split_run <- as.character(unique(tempted_metrics$split_run)[1])

mefisto_candidates <- sort(list.dirs(file.path(root, "data", "mefisto"), recursive = FALSE), decreasing = TRUE)
mefisto_run <- NA_character_
for (candidate in mefisto_candidates) {
  metrics_path <- file.path(candidate, "all_metrics.csv")
  if (!file.exists(metrics_path) || !run_is_complete(candidate, "MEFISTO")) next
  candidate_metrics <- read_table(metrics_path)
  if ("split_run" %in% names(candidate_metrics) && length(unique(candidate_metrics$split_run)) == 1 &&
      as.character(unique(candidate_metrics$split_run)[1]) == split_run) {
    mefisto_run <- candidate
    break
  }
}
if (is.na(mefisto_run)) stop("No complete five-factor MEFISTO run uses the same split set as TEMPTED")

batch_names <- intersect(
  basename(list.dirs(tempted_run, recursive = FALSE)),
  basename(list.dirs(mefisto_run, recursive = FALSE))
)
batch_names <- sort(batch_names[grepl("^batch_[0-9]+$", batch_names)])
if (length(batch_names) != EXPECTED_BATCHES) {
  stop(paste("Expected", EXPECTED_BATCHES, "matched model batches; found", length(batch_names)))
}

batch <- if (nzchar(SELECTED_BATCH)) SELECTED_BATCH else batch_names[1]
tempted_batch <- file.path(tempted_run, batch)
mefisto_batch <- file.path(mefisto_run, batch)
splits_run <- file.path(root, "data", "splits", split_run)
if (!dir.exists(splits_run)) stop(paste("Split run not found:", splits_run))
split_batch <- file.path(splits_run, batch)

# Verify exact held-out subject identity in every paired batch.
for (batch_name in batch_names) {
  t <- read_table(find_table(file.path(tempted_run, batch_name), "predictions"))
  m <- read_table(find_table(file.path(mefisto_run, batch_name), "predictions"))
  t <- t[order(as.character(t$subject_id)), c("subject_id", "truth")]
  m <- m[order(as.character(m$subject_id)), c("subject_id", "truth")]
  if (!identical(as.character(t$subject_id), as.character(m$subject_id)) ||
      !identical(as.character(t$truth), as.character(m$truth))) {
    stop(paste(batch_name, "does not contain identical TEMPTED/MEFISTO held-out subjects"))
  }
}

run_id <- format(Sys.time(), "%Y%m%d_%H%M%S")
data_output <- file.path(root, "data", "plot_data", run_id)
figure_output <- file.path(root, "data", "figures", run_id)
dir.create(data_output, recursive = TRUE)
dir.create(figure_output, recursive = TRUE)

cat(
  "Split:", splits_run,
  "\nTEMPTED:", tempted_run,
  "\nMEFISTO:", mefisto_run,
  "\nBatch:", batch,
  "\nPlot data:", data_output,
  "\nFigures:", figure_output, "\n"
)
```

```r
train_counts <- read_table(find_table(split_batch, "train_counts"))
test_counts <- read_table(find_table(split_batch, "test_counts"))
train_meta <- read_table(find_table(split_batch, "train_metadata"))
test_meta <- read_table(find_table(split_batch, "test_metadata"))

counts <- bind_rows(train_counts, test_counts)
metadata <- bind_rows(train_meta, test_meta) %>%
  mutate(
    sample_id = as.character(sample_id),
    subject_id = as.character(subject_id),
    label = factor(normalize_group(label), levels = GROUP_ORDER),
    time = as.numeric(time)
  )

rownames(counts) <- as.character(counts$sample_id)
counts$sample_id <- NULL
counts <- as.matrix(counts[metadata$sample_id, , drop = FALSE])
storage.mode(counts) <- "numeric"

read_matrix_table <- function(folder, stem) {
  path <- find_table(folder, stem)
  if (is.na(path)) return(NULL)
  x <- read_table(path)
  rownames(x) <- as.character(x$sample_id)
  x$sample_id <- NULL
  as.matrix(x)
}

# TEMPTED sample projections use the exact transformed matrix used by TEMPTED.
read_tempted_matrix <- function(folder, stem) {
  preferred <- read_matrix_table(folder, paste0(stem, "_tempted_clr"))
  if (!is.null(preferred)) return(preferred)

  # Backward compatibility for the older completed-rCLR split set.
  old <- read_matrix_table(folder, paste0(stem, "_rclr"))
  if (!is.null(old)) return(old)

  stop(paste("No TEMPTED transformed matrix found in", folder))
}

train_tempted <- read_tempted_matrix(split_batch, "train")
test_tempted <- read_tempted_matrix(split_batch, "test")
tempted_values <- rbind(train_tempted, test_tempted)
tempted_values <- tempted_values[metadata$sample_id, , drop = FALSE]

load_subject_scores <- function(folder) {
  x <- read_table(find_table(folder, "test_subject_scores"))
  factors <- factor_names(x)

  x %>%
    mutate(
      subject_id = as.character(subject_id),
      label = factor(normalize_group(label), levels = GROUP_ORDER)
    ) %>%
    select(subject_id, label, all_of(factors))
}

tempted_loadings <- load_loadings(tempted_batch)
mefisto_loadings <- load_loadings(mefisto_batch)
tempted_scores <- load_subject_scores(tempted_batch)
mefisto_scores <- load_subject_scores(mefisto_batch)

# Match TEMPTED components to MEFISTO factors using all 20 batches, not one split.
# First align factor order/sign within each method to that method's batch_001 reference.
align_loadings <- function(reference, current) {
  factors <- factor_names(reference)
  current_factors <- factor_names(current)
  shared <- intersect(reference$feature_id, current$feature_id)
  correlation <- matrix(NA_real_, length(factors), length(current_factors))
  for (i in seq_along(factors)) {
    x <- reference[[factors[i]]][match(shared, reference$feature_id)]
    for (j in seq_along(current_factors)) {
      y <- current[[current_factors[j]]][match(shared, current$feature_id)]
      correlation[i, j] <- suppressWarnings(cor(x, y, use = "complete.obs"))
    }
  }
  grid <- expand.grid(rep(list(seq_along(current_factors)), length(factors)))
  permutations <- grid[apply(grid, 1, function(x) length(unique(x)) == length(x)), , drop = FALSE]
  assignment_scores <- apply(permutations, 1, function(p) {
    values <- correlation[cbind(seq_along(factors), as.integer(p))]
    if (any(!is.finite(values))) return(-Inf)
    sum(abs(values))
  })
  best <- as.integer(permutations[which.max(assignment_scores), ])
  aligned <- data.frame(feature_id = current$feature_id, check.names = FALSE)
  for (i in seq_along(factors)) {
    sign_value <- ifelse(correlation[i, best[i]] < 0, -1, 1)
    aligned[[factors[i]]] <- current[[current_factors[best[i]]]] * sign_value
  }
  aligned
}

tempted_reference <- load_loadings(file.path(tempted_run, batch_names[1]))
mefisto_reference <- load_loadings(file.path(mefisto_run, batch_names[1]))
tempted_aligned <- list()
mefisto_aligned <- list()
for (batch_name in batch_names) {
  tempted_aligned[[batch_name]] <- align_loadings(tempted_reference, load_loadings(file.path(tempted_run, batch_name)))
  mefisto_aligned[[batch_name]] <- align_loadings(mefisto_reference, load_loadings(file.path(mefisto_run, batch_name)))
}

tempted_factors <- factor_names(tempted_reference)
mefisto_factors <- factor_names(mefisto_reference)
correlation <- matrix(NA_real_, length(tempted_factors), length(mefisto_factors))
for (i in seq_along(tempted_factors)) {
  for (j in seq_along(mefisto_factors)) {
    batch_correlations <- c()
    for (batch_name in batch_names) {
      t <- tempted_aligned[[batch_name]]
      m <- mefisto_aligned[[batch_name]]
      shared <- intersect(t$feature_id, m$feature_id)
      r <- suppressWarnings(cor(
        t[[tempted_factors[i]]][match(shared, t$feature_id)],
        m[[mefisto_factors[j]]][match(shared, m$feature_id)],
        use = "complete.obs"
      ))
      if (is.finite(r)) batch_correlations <- c(batch_correlations, r)
    }
    if (length(batch_correlations)) correlation[i, j] <- median(batch_correlations)
  }
}

grid <- expand.grid(rep(list(seq_along(mefisto_factors)), length(tempted_factors)))
permutations <- grid[apply(grid, 1, function(x) length(unique(x)) == length(x)), , drop = FALSE]
scores <- apply(permutations, 1, function(p) {
  values <- correlation[cbind(seq_along(tempted_factors), as.integer(p))]
  if (any(!is.finite(values))) return(-Inf)
  sum(abs(values))
})
best <- as.integer(permutations[which.max(scores), ])

alignment <- data.frame(
  tempted_component = tempted_factors,
  mefisto_factor = mefisto_factors[best],
  loading_correlation = correlation[cbind(seq_along(tempted_factors), best)]
)
alignment$sign <- ifelse(alignment$loading_correlation < 0, -1, 1)

for (row in seq_len(nrow(alignment))) {
  factor_name <- alignment$mefisto_factor[row]
  sign_value <- alignment$sign[row]
  mefisto_loadings[[factor_name]] <- mefisto_loadings[[factor_name]] * sign_value
  mefisto_scores[[factor_name]] <- mefisto_scores[[factor_name]] * sign_value
}

subject_scores <- bind_rows(
  tempted_scores %>% mutate(method = "TEMPTED"),
  mefisto_scores %>% mutate(method = "MEFISTO")
)

project_tempted_samples <- function(loadings) {
  factors <- factor_names(loadings)
  shared <- intersect(loadings$feature_id, colnames(tempted_values))
  W <- as.matrix(loadings[match(shared, loadings$feature_id), factors, drop = FALSE])
  X <- tempted_values[, shared, drop = FALSE]
  projected <- X %*% W

  data.frame(sample_id = metadata$sample_id, projected, check.names = FALSE) %>%
    bind_cols(metadata %>% select(subject_id, time, label)) %>%
    mutate(method = "TEMPTED")
}

load_mefisto_samples <- function(folder) {
  train <- read_table(find_table(folder, "train_sample_factors"))
  test <- read_table(find_table(folder, "test_sample_factors"))
  bind_rows(train, test) %>%
    mutate(
      sample_id = as.character(sample_id),
      subject_id = as.character(subject_id),
      time = as.numeric(time),
      label = factor(normalize_group(label), levels = GROUP_ORDER),
      method = "MEFISTO"
    ) %>%
    select(sample_id, all_of(mefisto_factors), subject_id, time, label, method)
}

mefisto_samples <- load_mefisto_samples(mefisto_batch)
for (row in seq_len(nrow(alignment))) {
  factor_name <- alignment$mefisto_factor[row]
  mefisto_samples[[factor_name]] <- mefisto_samples[[factor_name]] * alignment$sign[row]
}

sample_factors <- bind_rows(
  project_tempted_samples(tempted_loadings),
  mefisto_samples
)

feature_loadings <- bind_rows(
  tempted_loadings %>% mutate(method = "TEMPTED"),
  mefisto_loadings %>% mutate(method = "MEFISTO")
)

write.csv(alignment, file.path(data_output, "factor_alignment.csv"), row.names = FALSE)
```

```r
factor_names <- function(x) {
  grep(
    "^factor_[0-9]+$",
    names(x),
    value = TRUE
  )
}


# Align repeated loading estimates within each method, then align MEFISTO
# factor order/sign to TEMPTED so corresponding factor_N columns are comparable.

alignment_score <- function(reference, current) {
  reference_factors <- factor_names(reference)
  current_factors <- factor_names(current)
  shared <- intersect(reference$feature_id, current$feature_id)

  if (length(shared) < 10) return(-Inf)

  correlation <- matrix(
    NA_real_,
    nrow = length(reference_factors),
    ncol = length(current_factors)
  )

  for (i in seq_along(reference_factors)) {
    x <- reference[[reference_factors[i]]][
      match(shared, reference$feature_id)
    ]

    for (j in seq_along(current_factors)) {
      y <- current[[current_factors[j]]][
        match(shared, current$feature_id)
      ]

      correlation[i, j] <- suppressWarnings(
        cor(x, y, use = "complete.obs")
      )
    }
  }

  grid <- expand.grid(
    rep(list(seq_along(current_factors)), length(reference_factors))
  )

  permutations <- grid[
    apply(
      grid,
      1,
      function(x) length(unique(x)) == length(x)
    ),
    ,
    drop = FALSE
  ]

  scores <- apply(
    permutations,
    1,
    function(p) {
      values <- correlation[
        cbind(
          seq_along(reference_factors),
          as.integer(p)
        )
      ]

      if (any(!is.finite(values))) return(-Inf)

      sum(abs(values))
    }
  )

  max(scores)
}


align_to_reference <- function(reference, current) {
  reference_factors <- factor_names(reference)
  current_factors <- factor_names(current)
  shared <- intersect(reference$feature_id, current$feature_id)

  if (length(shared) < 10) {
    stop("Too few shared genera to align loading matrices.")
  }

  correlation <- matrix(
    NA_real_,
    nrow = length(reference_factors),
    ncol = length(current_factors)
  )

  for (i in seq_along(reference_factors)) {
    x <- reference[[reference_factors[i]]][
      match(shared, reference$feature_id)
    ]

    for (j in seq_along(current_factors)) {
      y <- current[[current_factors[j]]][
        match(shared, current$feature_id)
      ]

      correlation[i, j] <- suppressWarnings(
        cor(x, y, use = "complete.obs")
      )
    }
  }

  grid <- expand.grid(
    rep(list(seq_along(current_factors)), length(reference_factors))
  )

  permutations <- grid[
    apply(
      grid,
      1,
      function(x) length(unique(x)) == length(x)
    ),
    ,
    drop = FALSE
  ]

  scores <- apply(
    permutations,
    1,
    function(p) {
      values <- correlation[
        cbind(
          seq_along(reference_factors),
          as.integer(p)
        )
      ]

      if (any(!is.finite(values))) return(-Inf)

      sum(abs(values))
    }
  )

  best <- as.integer(
    permutations[which.max(scores), ]
  )

  aligned <- data.frame(
    feature_id = current$feature_id,
    stringsAsFactors = FALSE
  )

  for (i in seq_along(reference_factors)) {
    j <- best[i]

    sign_value <- ifelse(
      correlation[i, j] < 0,
      -1,
      1
    )

    aligned[[reference_factors[i]]] <-
      current[[current_factors[j]]] * sign_value
  }

  aligned
}


# Load all batches first.

method_loadings <- list()

for (method_name in c("TEMPTED", "MEFISTO")) {

  run_folder <- if (method_name == "TEMPTED") {
    tempted_run
  } else {
    mefisto_run
  }

  method_loadings[[method_name]] <- lapply(
    batch_names,
    function(batch_name) {
      load_loadings(
        file.path(run_folder, batch_name)
      )
    }
  )

  names(method_loadings[[method_name]]) <- batch_names
}


# Choose the most representative batch within each method.
# The reference is the batch with the highest average alignment score
# against all other batches.

representative_batch <- function(loadings) {

  scores <- sapply(
    seq_along(loadings),
    function(i) {

      other_scores <- sapply(
        seq_along(loadings),
        function(j) {

          if (i == j) return(NA_real_)

          alignment_score(
            loadings[[i]],
            loadings[[j]]
          )
        }
      )

      mean(other_scores, na.rm = TRUE)
    }
  )

  which.max(scores)
}


tempted_reference_index <-
  representative_batch(method_loadings$TEMPTED)

mefisto_reference_index <-
  representative_batch(method_loadings$MEFISTO)

tempted_reference <-
  method_loadings$TEMPTED[[tempted_reference_index]]

mefisto_reference <-
  method_loadings$MEFISTO[[mefisto_reference_index]]

cat(
  "TEMPTED reference:",
  batch_names[tempted_reference_index],
  "\n"
)

cat(
  "MEFISTO reference:",
  batch_names[mefisto_reference_index],
  "\n"
)


# Align every batch to its own method's representative reference.

aligned_by_method <- list()

for (method_name in c("TEMPTED", "MEFISTO")) {

  reference <- if (method_name == "TEMPTED") {
    tempted_reference
  } else {
    mefisto_reference
  }

  aligned_by_method[[method_name]] <- lapply(
    method_loadings[[method_name]],
    function(current) {
      align_to_reference(
        reference,
        current
      )
    }
  )

  names(aligned_by_method[[method_name]]) <-
    batch_names
}


# Calculate median within-method loadings first.
# These are much more stable than using one arbitrary batch for
# TEMPTED-versus-MEFISTO matching.

median_loadings <- function(loadings) {

  combined <- bind_rows(
    Map(
      function(x, batch_name) {
        x$batch <- batch_name
        x
      },
      loadings,
      names(loadings)
    )
  )

  combined %>%
    group_by(feature_id) %>%
    summarise(
      across(
        matches("^factor_[0-9]+$"),
        ~ median(.x, na.rm = TRUE)
      ),
      .groups = "drop"
    )
}


tempted_median <-
  median_loadings(aligned_by_method$TEMPTED)

mefisto_median <-
  median_loadings(aligned_by_method$MEFISTO)


# Match the five consensus MEFISTO factors to the five
# consensus TEMPTED components.

tempted_factors <- factor_names(tempted_median)
mefisto_factors <- factor_names(mefisto_median)

shared <- intersect(
  tempted_median$feature_id,
  mefisto_median$feature_id
)

cross_correlation <- matrix(
  NA_real_,
  nrow = length(tempted_factors),
  ncol = length(mefisto_factors)
)

for (i in seq_along(tempted_factors)) {

  x <- tempted_median[[tempted_factors[i]]][
    match(shared, tempted_median$feature_id)
  ]

  for (j in seq_along(mefisto_factors)) {

    y <- mefisto_median[[mefisto_factors[j]]][
      match(shared, mefisto_median$feature_id)
    ]

    cross_correlation[i, j] <- suppressWarnings(
      cor(x, y, use = "complete.obs")
    )
  }
}


grid <- expand.grid(
  rep(
    list(seq_along(mefisto_factors)),
    length(tempted_factors)
  )
)

permutations <- grid[
  apply(
    grid,
    1,
    function(x) length(unique(x)) == length(x)
  ),
  ,
  drop = FALSE
]

scores <- apply(
  permutations,
  1,
  function(p) {

    values <- cross_correlation[
      cbind(
        seq_along(tempted_factors),
        as.integer(p)
      )
    ]

    if (any(!is.finite(values))) return(-Inf)

    sum(abs(values))
  }
)

best <- as.integer(
  permutations[which.max(scores), ]
)


alignment <- data.frame(
  tempted_factor = tempted_factors,
  mefisto_factor = mefisto_factors[best],
  correlation = cross_correlation[
    cbind(
      seq_along(tempted_factors),
      best
    )
  ],
  stringsAsFactors = FALSE
)

alignment$sign <- ifelse(
  alignment$correlation < 0,
  -1,
  1
)

print(alignment)


# Actually reorder AND sign-correct every MEFISTO loading table
# into TEMPTED factor order.

aligned_mefisto <- lapply(
  aligned_by_method$MEFISTO,
  function(current) {

    result <- data.frame(
      feature_id = current$feature_id,
      stringsAsFactors = FALSE
    )

    for (i in seq_along(tempted_factors)) {

      source_factor <-
        alignment$mefisto_factor[i]

      target_factor <-
        tempted_factors[i]

      result[[target_factor]] <-
        current[[source_factor]] *
        alignment$sign[i]
    }

    result
  }
)

names(aligned_mefisto) <- batch_names


# Combine all 20 aligned loading estimates.

loading_resample_rows <- list()
index <- 1

for (batch_name in batch_names) {

  tempted_current <-
    aligned_by_method$TEMPTED[[batch_name]]

  tempted_current$method <- "TEMPTED"
  tempted_current$batch <- batch_name

  loading_resample_rows[[index]] <-
    tempted_current

  index <- index + 1


  mefisto_current <-
    aligned_mefisto[[batch_name]]

  mefisto_current$method <- "MEFISTO"
  mefisto_current$batch <- batch_name

  loading_resample_rows[[index]] <-
    mefisto_current

  index <- index + 1
}


loading_resamples <-
  bind_rows(loading_resample_rows)


# Median aligned loading across all 20 batches.

feature_loadings <- loading_resamples %>%
  group_by(method, feature_id) %>%
  summarise(
    across(
      matches("^factor_[0-9]+$"),
      ~ median(.x, na.rm = TRUE)
    ),
    .groups = "drop"
  )


write.csv(
  subject_scores,
  file.path(
    data_output,
    "subject_scores.csv"
  ),
  row.names = FALSE
)

write.csv(
  sample_factors,
  file.path(
    data_output,
    "sample_factors.csv"
  ),
  row.names = FALSE
)

write.csv(
  feature_loadings,
  file.path(
    data_output,
    "feature_loadings.csv"
  ),
  row.names = FALSE
)

write.csv(
  trajectory_summary,
  file.path(
    data_output,
    "trajectory_summary.csv"
  ),
  row.names = FALSE
)

write.csv(
  loading_resamples,
  file.path(
    data_output,
    "loading_resamples.csv"
  ),
  row.names = FALSE
)

write.csv(
  alignment,
  file.path(
    data_output,
    "factor_alignment.csv"
  ),
  row.names = FALSE
)


source_manifest <- data.frame(
  item = c(
    "splits_run",
    "split_run",
    "tempted_run",
    "mefisto_run",
    "batch",
    "batches",
    "dimensions",
    "tempted_reference_batch",
    "mefisto_reference_batch"
  ),
  value = c(
    splits_run,
    split_run,
    tempted_run,
    mefisto_run,
    batch,
    EXPECTED_BATCHES,
    EXPECTED_DIMENSIONS,
    batch_names[tempted_reference_index],
    batch_names[mefisto_reference_index]
  )
)

write.csv(
  source_manifest,
  file.path(
    data_output,
    "source_manifest.csv"
  ),
  row.names = FALSE
)

cat(
  "Notebook 06 finished.",
  "\nPlot data:",
  data_output,
  "\nFigures:",
  figure_output,
  "\n"
)
```
