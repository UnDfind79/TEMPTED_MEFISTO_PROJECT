# 08 — TEMPTED component animations

This notebook creates separate TEMPTED subject and feature animations.

* **Primary subject animations:** Components 1 and 2 are sorted by `num_abx_treatments`, and `num_abx_treatments` is the only metadata-strip column.
* **Additional Component 2 subject animation:** sorted by `num_abx_first_year`, with `num_abx_first_year` as the only metadata-strip column.
* **Feature animation:** $B_{jr}\phi_r(t)$ for the strongest negative and positive feature loadings.
* **Time:** all animations retain the same $\phi_r(t)$ time window along the bottom.

Component signs preserve the orientation saved by the TEMPTED run / notebook 6.


```python
from datetime import datetime
from pathlib import Path
import gc
import shutil

import matplotlib.pyplot as plt
from matplotlib import colors as mcolors
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Patch
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import pandas as pd
from IPython.display import Image, display

# animation settings
N_FEATURES_EACH_SIDE = 10
ASSOCIATION_TOP_N = 3
METADATA_Z_CLIP = 2.0
FPS = 12
DPI = 140
FIGSIZE = (13, 10)

# sort subject rows by antibiotic-treatment metadata rather than TEMPTED subject score
SUBJECT_SORT_METADATA = "num_abx_treatments"


root = Path(".") if Path("data").exists() else Path("..")
t_dir = sorted((root / "data" / "tempted").iterdir())[-1]
meta_dir = sorted((root / "data" / "metadata_analysis").iterdir())[-1]

output = root / "data" / "tempted_animations" / datetime.now().strftime("%Y%m%d_%H%M%S")
output.mkdir(parents=True)



# use sored metadata/tempted table
subjects_with_metadata = pd.read_csv(
    meta_dir / "tempted_with_metadata.csv",
    dtype={"subject_id": str},
)
metadata_associations = pd.read_csv(meta_dir / "metadata_associations.csv")

feature_loadings = pd.read_csv(t_dir / "feature_loadings.csv")
temporal_loadings = pd.read_csv(t_dir / "temporal_loadings.csv")


components = [c for c in temporal_loadings.columns if c.startswith("component_")]

countries = ["FIN", "EST", "RUS"]
colors = {
    "FIN": "green",
    "EST": "blue",
    "RUS": "red",
}

print("TEMPTED run:", t_dir)
print("Metadata analysis run:", meta_dir)
print("Components:", components)
print("Animation output:", output)
```

```python
# helper functions

def symmetric_limit(values, pad=1.08):
    """Give a zero-centered axis a little space beyond its largest value."""
    return np.max(np.abs(values)) * pad


def component_orientation(component):
    """Make phi positive where its absolute value is largest."""
    phi = temporal_loadings[component].to_numpy(float)
    peak = np.argmax(np.abs(phi))
    return 1.0 if phi[peak] > 0 else -1.0


def component_associations(component, orientation):
    """Return the strongest TEMPTED metadata associations in the plot orientation."""
    table = metadata_associations[
        (metadata_associations["method"] == "TEMPTED")
        & (metadata_associations["dimension"] == component)
    ].copy()

    table["association"] = table["association"] * orientation
    table["absolute_association"] = table["association"].abs()

    return (
        table
        .sort_values("absolute_association", ascending=False)
        .head(ASSOCIATION_TOP_N)
        .reset_index(drop=True)
    )
```

```python
# build and save separate subject and feature animations for every component

outputs = []

# Primary subject plots for Components 1 and 2:
#   * sorted by num_abx_treatments
#   * num_abx_treatments is the ONLY metadata strip column
PRIMARY_SUBJECT_METADATA = "num_abx_treatments"

# Additional Component 2 subject plot:
#   * sorted by the number of antibiotic treatments in the first year
#   * num_abx_first_year is the ONLY metadata strip column
FIRST_YEAR_ABX_METADATA = "num_abx_first_year"


def temporal_inflection_times(x, y):
    """Return time points where the temporal loading changes curvature."""
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    first_derivative = np.gradient(y, x)
    second_derivative = np.gradient(first_derivative, x)

    signs = np.sign(second_derivative)

    # Carry neighboring nonzero signs through exact numerical zeros so a
    # zero-valued second derivative does not create duplicate crossings.
    for index in range(1, len(signs)):
        if signs[index] == 0:
            signs[index] = signs[index - 1]

    for index in range(len(signs) - 2, -1, -1):
        if signs[index] == 0:
            signs[index] = signs[index + 1]

    crossings = np.where(signs[:-1] * signs[1:] < 0)[0] + 1
    return x[crossings]


def configure_time_axis(ax_time, time, phi, component):
    """Draw the common temporal-loading window used by both animation types."""
    time_limit = symmetric_limit(phi)

    ax_time.plot(
        time,
        phi,
        color="blue",
        linewidth=1.7,
        zorder=5,
    )

    for inflection_time in temporal_inflection_times(time, phi):
        ax_time.axvline(
            inflection_time,
            color="black",
            linewidth=0.9,
            alpha=0.55,
            zorder=3,
        )

    ax_time.axhline(
        0,
        color="black",
        linewidth=0.7,
        alpha=0.6,
        zorder=4,
    )

    marker, = ax_time.plot(
        [time[0]],
        [phi[0]],
        "o",
        color="red",
        markersize=6,
        zorder=6,
    )

    ax_time.set_xlim(time.min(), time.max())
    ax_time.set_ylim(-time_limit, time_limit)
    ax_time.set_xlabel("Age")
    ax_time.set_ylabel(r"$\phi(t)$")
    ax_time.set_title(
        f"{component}: temporal loading",
        fontsize=14,
    )
    ax_time.grid(alpha=0.15)

    return marker


def standardized_metadata(values):
    """Standardize one metadata vector for display in the black-white strip."""
    numeric = pd.to_numeric(values, errors="coerce").astype(float)

    mean = numeric.mean()
    std = numeric.std(ddof=0)

    if pd.isna(std) or std == 0:
        z = pd.Series(
            np.where(numeric.notna(), 0.0, np.nan),
            index=numeric.index,
            dtype=float,
        )
    else:
        z = (numeric - mean) / std

    return np.clip(
        z.to_numpy(float) / METADATA_Z_CLIP,
        -1,
        1,
    )


def build_subject_animation(
    component,
    time,
    phi,
    metadata_name,
    output_path,
    plot_label,
):
    """Build one subject animation sorted by one displayed metadata variable."""

    # The displayed metadata variable is also the sorting variable.
    subject_columns = [
        "subject_id",
        "country",
        component,
        metadata_name,
    ]

    subjects = subjects_with_metadata[
        subject_columns
    ].copy()

    # Preserve the original TEMPTED sign convention used in notebook 6.
    subjects[component] = subjects[component].astype(float)

    # barh/imshow put row 0 at the bottom. Ascending order therefore places
    # the largest metadata values at the top. Missing values are placed first,
    # which means they remain below the observed values in the figure.
    subjects["_sort_value"] = pd.to_numeric(
        subjects[metadata_name],
        errors="coerce",
    )

    subjects = (
        subjects
        .sort_values(
            "_sort_value",
            ascending=True,
            na_position="first",
        )
        .drop(columns="_sort_value")
        .reset_index(drop=True)
    )

    metadata_matrix = standardized_metadata(
        subjects[metadata_name]
    )[:, None]

    subject_base = subjects[component].to_numpy(float)
    subject_colors = [
        colors[country]
        for country in subjects["country"]
    ]

    phi_max = np.max(np.abs(phi))
    subject_limit = symmetric_limit(
        subject_base * phi_max
    )

    fig = plt.figure(figsize=FIGSIZE)

    outer = fig.add_gridspec(
        2, 1,
        height_ratios=[4.8, 1.20],
        left=0.08,
        right=0.97,
        bottom=0.12,
        top=0.90,
        hspace=0.30,
    )

    # Exactly one narrow metadata column.
    subject_grid = outer[0, 0].subgridspec(
        1, 2,
        width_ratios=[0.075, 1.00],
        wspace=0.01,
    )

    ax_strip = fig.add_subplot(subject_grid[0, 0])
    ax_subject = fig.add_subplot(
        subject_grid[0, 1],
        sharey=ax_strip,
    )
    ax_time = fig.add_subplot(outer[1, 0])

    subject_y = np.arange(len(subjects))

    cmap = LinearSegmentedColormap.from_list(
        "metadata",
        ["white", "darkgray", "black"],
    )
    cmap.set_bad("orange")

    ax_strip.imshow(
        np.ma.masked_invalid(metadata_matrix),
        aspect="auto",
        origin="lower",
        interpolation="nearest",
        cmap=cmap,
        norm=mcolors.TwoSlopeNorm(
            vmin=-1,
            vcenter=0,
            vmax=1,
        ),
        extent=(
            -0.5,
            0.5,
            -0.5,
            len(subjects) - 0.5,
        ),
    )

    ax_strip.set_xlim(-0.5, 0.5)
    ax_strip.set_ylim(
        -0.5,
        len(subjects) - 0.5,
    )
    ax_strip.set_xticks([0])

    metadata_label = str(
        metadata_name
    ).replace("_", " ")

    ax_strip.set_xticklabels(
        [metadata_label],
        rotation=30,
        ha="left",
        va="bottom",
        fontsize=14,
    )

    ax_strip.tick_params(
        axis="x",
        top=True,
        labeltop=True,
        bottom=False,
        labelbottom=False,
        pad=2,
    )
    ax_strip.set_yticks([])

    subject_bars = ax_subject.barh(
        subject_y,
        subject_base * phi[0],
        color=subject_colors,
        alpha=0.82,
    )

    ax_subject.axvline(
        0,
        color="black",
        linewidth=1,
    )
    ax_subject.set_xlim(
        -subject_limit,
        subject_limit,
    )
    ax_subject.set_yticks([])
    ax_subject.set_xlabel(
        r"Time-scaled subject contribution  $A_i\phi(t)$"
    )
    ax_subject.set_title("Subjects")
    ax_subject.grid(
        axis="x",
        alpha=0.15,
    )

    marker_subject = configure_time_axis(
        ax_time,
        time,
        phi,
        component,
    )

    country_handles = [
        plt.Line2D(
            [0],
            [0],
            marker="s",
            linestyle="",
            markersize=7,
            markerfacecolor=colors[country],
            markeredgecolor=colors[country],
            label=country,
        )
        for country in countries
    ]

    country_legend = ax_subject.legend(
        handles=country_handles,
        title="Country",
        loc="lower right",
        fontsize=12,
        title_fontsize=14,
    )
    ax_subject.add_artist(country_legend)

    strip_handles = [
        Patch(
            facecolor=cmap(1.0),
            label="High",
        ),
        Patch(
            facecolor=cmap(0.5),
            label="Average",
        ),
        Patch(
            facecolor=cmap(0.0),
            label="Low",
        ),
        Patch(
            facecolor="orange",
            label="Missing",
        ),
    ]

    ax_subject.legend(
        handles=strip_handles,
        title="Metadata values",
        loc="upper right",
        fontsize=12,
        title_fontsize=14,
    )

    fig.text(
        0.08,
        0.045,
        f"Subjects sorted by {metadata_label}",
        ha="left",
        va="center",
        fontsize=12,
    )

    subject_frame_label = fig.text(
        0.5,
        0.965,
        "",
        ha="center",
        va="top",
        fontsize=12,
    )

    def update_subject(frame):
        p = phi[frame]

        for bar, value in zip(
            subject_bars,
            subject_base * p,
        ):
            bar.set_width(value)

        marker_subject.set_data(
            [time[frame]],
            [p],
        )

        subject_frame_label.set_text(
            f"{component} | {plot_label} | "
            f"age = {time[frame]:.1f} | "
            f"temporal loading = {p:.3f}"
        )

        return [
            *subject_bars,
            marker_subject,
            subject_frame_label,
        ]

    subject_animation = FuncAnimation(
        fig,
        update_subject,
        frames=len(time),
        interval=1000 / FPS,
        blit=False,
        repeat=True,
        cache_frame_data=False,
    )

    subject_animation.save(
        output_path,
        writer=PillowWriter(fps=FPS),
        dpi=DPI,
    )

    plt.close(fig)
    print("Saved:", output_path)

    del subject_animation
    gc.collect()


for component in components:
    time = temporal_loadings[
        "time"
    ].to_numpy(float)

    # Preserve the original TEMPTED component orientation exactly as saved
    # and displayed in notebook 6. Do not independently sign-flip components.
    phi = temporal_loadings[
        component
    ].to_numpy(float)

    # ------------------------------------------------------------------
    # PRIMARY SUBJECT ANIMATION
    # Components 1 and 2 are sorted by num_abx_treatments and show only
    # num_abx_treatments in the metadata strip.
    # ------------------------------------------------------------------
    subject_path = (
        output
        / f"{component}_tempted_subjects.gif"
    )

    build_subject_animation(
        component=component,
        time=time,
        phi=phi,
        metadata_name=PRIMARY_SUBJECT_METADATA,
        output_path=subject_path,
        plot_label="subjects | num abx",
    )

    # ------------------------------------------------------------------
    # SECOND COMPONENT 2 SUBJECT ANIMATION
    # Sorted by num_abx_first_year and showing only that metadata column.
    # ------------------------------------------------------------------
    first_year_subject_path = None

    if component == "component_2":
        first_year_subject_path = (
            output
            / "component_2_tempted_subjects_first_year_abx.gif"
        )

        build_subject_animation(
            component=component,
            time=time,
            phi=phi,
            metadata_name=FIRST_YEAR_ABX_METADATA,
            output_path=first_year_subject_path,
            plot_label="subjects | num abx first year",
        )

    # ------------------------------------------------------------------
    # FEATURE-LOADING ANIMATION
    # ------------------------------------------------------------------
    features = pd.concat([
        feature_loadings.nsmallest(
            N_FEATURES_EACH_SIDE,
            component,
        ),
        feature_loadings.nlargest(
            N_FEATURES_EACH_SIDE,
            component,
        ),
    ])

    features = (
        features
        .drop_duplicates()
        .sort_values(component)
        .reset_index(drop=True)
    )

    feature_labels = (
        features["feature_id"]
        .astype(str)
        .tolist()
    )

    feature_base = features[
        component
    ].to_numpy(float)

    phi_max = np.max(np.abs(phi))
    feature_limit = symmetric_limit(
        feature_base * phi_max
    )

    fig = plt.figure(figsize=FIGSIZE)

    outer = fig.add_gridspec(
        2, 1,
        height_ratios=[4.8, 1.20],
        left=0.23,
        right=0.97,
        bottom=0.12,
        top=0.90,
        hspace=0.30,
    )

    ax_feature = fig.add_subplot(
        outer[0, 0]
    )
    ax_time = fig.add_subplot(
        outer[1, 0]
    )

    feature_y = np.arange(
        len(features)
    )

    feature_bars = ax_feature.barh(
        feature_y,
        feature_base * phi[0],
        color="0.45",
        alpha=0.85,
    )

    ax_feature.axvline(
        0,
        color="black",
        linewidth=1,
    )
    ax_feature.set_xlim(
        -feature_limit,
        feature_limit,
    )
    ax_feature.set_yticks(
        feature_y
    )
    ax_feature.set_yticklabels(
        feature_labels,
        fontsize=12,
    )
    ax_feature.set_xlabel(
        r"Time-scaled feature contribution  $B_j\phi(t)$"
    )
    ax_feature.set_title(
        "Feature loadings"
    )
    ax_feature.grid(
        axis="x",
        alpha=0.15,
    )

    marker_feature = configure_time_axis(
        ax_time,
        time,
        phi,
        component,
    )

    feature_frame_label = fig.text(
        0.5,
        0.965,
        "",
        ha="center",
        va="top",
        fontsize=12,
    )

    def update_feature(frame):
        p = phi[frame]

        for bar, value in zip(
            feature_bars,
            feature_base * p,
        ):
            bar.set_width(value)

        marker_feature.set_data(
            [time[frame]],
            [p],
        )

        feature_frame_label.set_text(
            f"{component} | features | "
            f"age = {time[frame]:.1f} | "
            f"temporal loading = {p:.3f}"
        )

        return [
            *feature_bars,
            marker_feature,
            feature_frame_label,
        ]

    feature_animation = FuncAnimation(
        fig,
        update_feature,
        frames=len(time),
        interval=1000 / FPS,
        blit=False,
        repeat=True,
        cache_frame_data=False,
    )

    feature_path = (
        output
        / f"{component}_tempted_features.gif"
    )

    feature_animation.save(
        feature_path,
        writer=PillowWriter(fps=FPS),
        dpi=DPI,
    )

    plt.close(fig)
    print("Saved:", feature_path)

    del feature_animation
    gc.collect()

    outputs.append({
        "component": component,
        "subject_animation": str(
            subject_path
        ),
        "first_year_abx_subject_animation": (
            str(first_year_subject_path)
            if first_year_subject_path is not None
            else None
        ),
        "feature_animation": str(
            feature_path
        ),
    })


pd.DataFrame(outputs)
```

```python
# preview every generated subject and feature animation

for row in outputs:
    component = row["component"]

    print(
        f"{component} — subjects sorted by num abx"
    )
    display(
        Image(
            filename=str(
                Path(row["subject_animation"])
            )
        )
    )

    first_year_path = row.get(
        "first_year_abx_subject_animation"
    )

    if first_year_path:
        print(
            f"{component} — subjects sorted by first year abx"
        )
        display(
            Image(
                filename=str(
                    Path(first_year_path)
                )
            )
        )

    print(
        f"{component} — feature loadings"
    )
    display(
        Image(
            filename=str(
                Path(row["feature_animation"])
            )
        )
    )
```
