# 05 — Attach metadata and find useful associations

TEMPTED gets metadata that are constant within subjects. MEFISTO gets the full sample-level metadata. Simple association scores identify useful plots.

This notebook treats metadata as **post-hoc interpretation** of the unsupervised latent factors rather than as labels used to fit the models. This matches the MOFA guidance to learn factors unsupervised and relate them to biological covariates afterward.

**Source:** MOFA FAQ, downstream analysis:  
https://biofam.github.io/MOFA2/faq.html

```python
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import spearmanr

root = Path(".") if Path("data").exists() else Path("..")
prep = sorted((root / "data" / "preprocessing").iterdir())[-1]
t_dir = sorted((root / "data" / "tempted").iterdir())[-1]
m_dir = sorted((root / "data" / "mefisto").iterdir())[-1]
output = root / "data" / "metadata_analysis" / datetime.now().strftime("%Y%m%d_%H%M%S")
output.mkdir(parents=True)

metadata = pd.read_csv(prep / "metadata.csv", dtype={"sample_id": str, "subject_id": str})
inventory = pd.read_csv(prep / "metadata_inventory.csv")
tempted = pd.read_csv(t_dir / "subject_scores.csv", dtype={"subject_id": str})
mefisto = pd.read_csv(m_dir / "sample_factors.csv", dtype={"sample_id": str, "subject_id": str})

t_dims = [c for c in tempted if c.startswith("component_")]
m_dims = [c for c in mefisto if c.startswith("factor_")]

constant = inventory.loc[inventory["subject_constant"], "metadata"]
subject_meta = metadata[["subject_id"] + [c for c in constant if c != "subject_id"]].drop_duplicates("subject_id")

tempted = tempted.merge(subject_meta, on="subject_id", how="left")
mefisto = mefisto.merge(metadata, on=["sample_id", "subject_id", "age"], how="left")

tempted.to_csv(output / "tempted_with_metadata.csv", index=False)
mefisto.to_csv(output / "mefisto_with_metadata.csv", index=False)
```

```python
# Automatically score usable metadata fields.
rows = []
ignore = {"sample_id", "subject_id", "SampleID", "subjectID", "library_size"}

for method, frame, dims in [("TEMPTED", tempted, t_dims), ("MEFISTO", mefisto, m_dims)]:
    meta = [c for c in frame if c not in ignore and c not in dims]

    for variable in meta:
        values = frame[variable]
        numeric = pd.to_numeric(values, errors="coerce")

        if numeric.notna().sum() >= 10 and numeric.nunique() >= 5:
            for dim in dims:
                keep = numeric.notna() & frame[dim].notna()
                rho = spearmanr(numeric[keep], frame.loc[keep, dim]).statistic
                rows.append([method, variable, "numeric", dim, rho, abs(rho), keep.sum()])
            continue

        counts = values.value_counts(dropna=True)
        groups = counts[counts >= 5].index
        if 2 <= len(groups) <= 8:
            temp = frame[values.isin(groups)]
            for dim in dims:
                z = (temp[dim] - temp[dim].mean()) / temp[dim].std(ddof=0)
                means = z.groupby(temp[variable]).mean()
                score = means.max() - means.min()
                rows.append([method, variable, "categorical", dim, score, abs(score), len(temp)])

associations = pd.DataFrame(
    rows,
    columns=["method", "metadata", "type", "dimension", "association", "absolute_association", "n"],
)
associations.to_csv(output / "metadata_associations.csv", index=False)

print("Saved:", output)
associations.sort_values("absolute_association", ascending=False).head(20)
```
