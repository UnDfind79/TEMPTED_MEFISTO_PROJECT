# 03 — Run TEMPTED

Fit TEMPTED once to the full filtered cohort. TEMPTED performs its own documented half-minimum pseudocount + CLR transformation.

```r
suppressPackageStartupMessages(library(tempted))

RANK <- 2
SMOOTH <- 1e-4
MAXITER <- 100
EPSILON <- 1e-4

root <- if (dir.exists("data")) "." else ".."
source <- tail(sort(list.dirs(file.path(root, "data", "preprocessing"), recursive = FALSE)), 1)
output <- file.path(root, "data", "tempted", format(Sys.time(), "%Y%m%d_%H%M%S"))
dir.create(output, recursive = TRUE)

counts <- read.csv(file.path(source, "counts_filtered.csv"), check.names = FALSE)
metadata <- read.csv(file.path(source, "metadata.csv"), stringsAsFactors = FALSE)

rownames(counts) <- counts$sample_id
counts$sample_id <- NULL
counts[] <- lapply(counts, as.numeric)
counts <- counts[metadata$sample_id, , drop = FALSE]

stopifnot(all(is.finite(as.matrix(counts))))
stopifnot(all(rowSums(counts > 0) > 0))
```

```r
# TEMPTED's default CLR uses half of each sample's smallest nonzero value.
x <- format_tempted(
  counts,
  metadata$age,
  metadata$subject_id,
  threshold = 1,
  pseudo = NULL,
  transform = "clr"
)

stopifnot(all(sapply(x, function(z) all(is.finite(z)))))

center <- svd_centralize(x, r = 1)
model <- tempted(
  center$datlist,
  r = RANK,
  smooth = SMOOTH,
  maxiter = MAXITER,
  epsilon = EPSILON
)
```

```r
scores <- as.data.frame(model$A_hat)
names(scores) <- paste0("component_", seq_len(ncol(scores)))
scores$subject_id <- rownames(scores)

loadings <- as.matrix(model$B_hat)
if (nrow(loadings) < ncol(loadings)) loadings <- t(loadings)

features <- rownames(loadings)
if (is.null(features)) features <- colnames(counts)

loadings <- data.frame(feature_id = features, loadings, check.names = FALSE)
names(loadings)[-1] <- paste0("component_", seq_len(ncol(loadings)))

write.csv(scores, file.path(output, "subject_scores.csv"), row.names = FALSE)
write.csv(loadings, file.path(output, "feature_loadings.csv"), row.names = FALSE)
saveRDS(model, file.path(output, "model.rds"))

cat("Saved", ncol(scores) - 1, "components to", output, "\n")
```
