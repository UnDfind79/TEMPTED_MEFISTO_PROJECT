# 02 — Common preprocessing and figures

**Input:** the cleaned cohort from notebook 01  
**Does:** filters genera once using the full cohort and creates shared preprocessing and metadata figures  
**Output:** one common filtered count table used by both methods, plus saved t-SNE/PCA coordinates, summary tables, and figures

Feature filtering is done **once on the full eligible cohort** because TEMPTED and MEFISTO are both fit before the later country-classification splits.

```python
from datetime import datetime
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

PREVALENCE = 0.10
MIN_RELATIVE_ABUNDANCE = 0.01
SEED = 2026
COUNTRIES = ["FIN", "EST", "RUS"]
COLORS = {"FIN": "#0072B2", "EST": "#009E73", "RUS": "#D55E00"}

root = Path(".") if Path("data").exists() else Path("..")
input_folder = sorted((root / "data" / "processed_16s").iterdir())[-1]
run_id = datetime.now().strftime("%Y%m%d_%H%M%S")
output = root / "data" / "preprocessing" / run_id
figures = output / "figures"
output.mkdir(parents=True, exist_ok=True)
figures.mkdir(parents=True, exist_ok=True)

counts = pd.read_csv(input_folder / "counts.csv", dtype={"sample_id": str}).set_index("sample_id")
metadata = pd.read_csv(input_folder / "metadata.csv", dtype={"sample_id": str, "subject_id": str, "country": str})
if "country" not in metadata.columns and "label" in metadata.columns:
    metadata = metadata.rename(columns={"label": "country"})
if "age" not in metadata.columns and "time" in metadata.columns:
    metadata = metadata.rename(columns={"time": "age"})
taxonomy_name = "taxonomy.csv" if (input_folder / "taxonomy.csv").exists() else "greengenes_genus_taxonomy.csv"
taxonomy = pd.read_csv(input_folder / taxonomy_name)

metadata["country"] = metadata["country"].replace({"FINLAND": "FIN", "ESTONIA": "EST", "RUSSIA": "RUS"})
metadata = metadata[metadata["country"].isin(COUNTRIES)].copy()
counts = counts.loc[metadata["sample_id"]]

print("Input:", input_folder)
print("Output:", output)
print("Samples:", len(metadata), "Subjects:", metadata["subject_id"].nunique(), "Genera:", counts.shape[1])
```

## Filter genera

The two filtering rules are written directly below so the retained feature set can be checked by hand.

```python
relative = counts.div(counts.sum(axis=1), axis=0)

prevalence = counts.gt(0).mean()
abundant = relative.ge(MIN_RELATIVE_ABUNDANCE).mean()

keep = (prevalence >= PREVALENCE) & (abundant >= PREVALENCE)
filtered = counts.loc[:, keep].copy()

summary = pd.DataFrame({
    "measure": [
        "samples",
        "subjects",
        "genera_before",
        "genera_retained",
        "genera_removed",
        "prevalence_cutoff",
        "relative_abundance_cutoff",
    ],
    "value": [
        len(metadata),
        metadata["subject_id"].nunique(),
        counts.shape[1],
        filtered.shape[1],
        counts.shape[1] - filtered.shape[1],
        PREVALENCE,
        MIN_RELATIVE_ABUNDANCE,
    ],
})
summary
```

## CLR, PCA, and t-SNE

The CLR transformation is kept explicit. One set of t-SNE coordinates is saved and reused for both country and age plots.

```python
x = filtered.astype(float).copy()

for sample in x.index:
    positive = x.loc[sample, x.loc[sample] > 0]
    pseudocount = positive.min() / 2
    x.loc[sample] = x.loc[sample] + pseudocount

clr = np.log(x)
clr = clr.sub(clr.mean(axis=1), axis=0)

if not np.isfinite(clr.to_numpy()).all():
    raise ValueError("CLR matrix contains non-finite values.")

pca = PCA(n_components=2, random_state=SEED)
pca_xy = pca.fit_transform(clr)

tsne = TSNE(n_components=2, random_state=SEED, init="pca", learning_rate="auto", perplexity=30)
tsne_xy = tsne.fit_transform(clr)

coordinates = metadata[["sample_id", "subject_id", "country", "age", "library_size"]].copy()
coordinates["pca_1"] = pca_xy[:, 0]
coordinates["pca_2"] = pca_xy[:, 1]
coordinates["tsne_1"] = tsne_xy[:, 0]
coordinates["tsne_2"] = tsne_xy[:, 1]

country_summary = (
    metadata.groupby("country")
    .agg(
        samples=("sample_id", "count"),
        subjects=("subject_id", pd.Series.nunique),
        mean_age=("age", "mean"),
        median_age=("age", "median"),
        mean_library_size=("library_size", "mean"),
    )
    .reindex(COUNTRIES)
    .reset_index()
)

visits = (
    metadata.groupby(["subject_id", "country"]).size().rename("visits").reset_index()
)
```

```python
filtered.reset_index().to_csv(output / "counts_filtered.csv", index=False)
metadata.to_csv(output / "metadata.csv", index=False)
taxonomy[taxonomy["feature"].isin(filtered.columns)].to_csv(output / "taxonomy.csv", index=False)
pd.DataFrame({"feature": filtered.columns}).to_csv(output / "retained_features.csv", index=False)
summary.to_csv(output / "preprocessing_summary.csv", index=False)
coordinates.to_csv(output / "coordinates.csv", index=False)
country_summary.to_csv(output / "country_summary.csv", index=False)
visits.to_csv(output / "visits_per_subject.csv", index=False)

print("Saved data to:", output)
```

## Preprocessing figures

The figures below emphasize country structure and relevant metadata while staying simple.

```python
def save(name):
    plt.tight_layout()
    plt.savefig(figures / name, dpi=220)
    plt.show()
    plt.close()

for xcol, ycol, name, title in [
    ("tsne_1", "tsne_2", "tsne_country.png", "t-SNE by country"),
    ("pca_1", "pca_2", "pca_country.png", "PCA by country"),
]:
    for country in COUNTRIES:
        rows = coordinates["country"] == country
        plt.scatter(coordinates.loc[rows, xcol], coordinates.loc[rows, ycol], s=18, alpha=0.7,
                    label=country, color=COLORS[country])
    plt.xlabel(xcol)
    plt.ylabel(ycol)
    plt.title(title)
    plt.legend(frameon=False)
    save(name)

for xcol, ycol, name, title in [
    ("tsne_1", "tsne_2", "tsne_age.png", "t-SNE by age at collection"),
    ("pca_1", "pca_2", "pca_age.png", "PCA by age at collection"),
]:
    plt.scatter(coordinates[xcol], coordinates[ycol], c=coordinates["age"], s=18, alpha=0.75, cmap="viridis")
    plt.xlabel(xcol)
    plt.ylabel(ycol)
    plt.title(title)
    plt.colorbar(label="Age at collection")
    save(name)
```

```python
metadata["country"].value_counts().reindex(COUNTRIES).plot.bar(color=[COLORS[x] for x in COUNTRIES])
plt.ylabel("Samples")
plt.title("Samples by country")
save("samples_by_country.png")

metadata.drop_duplicates("subject_id")["country"].value_counts().reindex(COUNTRIES).plot.bar(color=[COLORS[x] for x in COUNTRIES])
plt.ylabel("Subjects")
plt.title("Subjects by country")
save("subjects_by_country.png")

plt.boxplot([metadata.loc[metadata["country"] == c, "age"].dropna() for c in COUNTRIES], labels=COUNTRIES)
plt.ylabel("Age at collection")
plt.title("Age by country")
save("age_by_country.png")

plt.boxplot([metadata.loc[metadata["country"] == c, "library_size"].dropna() for c in COUNTRIES], labels=COUNTRIES)
plt.ylabel("Library size")
plt.title("Library size by country")
save("library_size_by_country.png")

plt.boxplot([visits.loc[visits["country"] == c, "visits"].dropna() for c in COUNTRIES], labels=COUNTRIES)
plt.ylabel("Samples per subject")
plt.title("Longitudinal sampling depth by country")
save("visits_by_country.png")
```

```python
summary.set_index("measure").loc[["genera_before", "genera_retained", "genera_removed"], "value"].plot.bar()
plt.ylabel("Genera")
plt.title("Feature filtering")
save("feature_filtering.png")

prevalence[filtered.columns].sort_values().plot.hist(bins=20)
plt.xlabel("Prevalence")
plt.title("Retained-genus prevalence")
save("genus_prevalence.png")

mean_abundance = relative[filtered.columns].mean().sort_values().tail(15)
labels = [name.split("|")[-1].replace("g__", "") for name in mean_abundance.index]
plt.barh(labels, mean_abundance.values)
plt.xlabel("Mean relative abundance")
plt.title("Top retained genera")
save("top_genera.png")

country_summary
```
