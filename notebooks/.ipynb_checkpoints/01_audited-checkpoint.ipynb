# 01 — Process DIABIMMUNE 16S data

**Input:** raw DIABIMMUNE 16S counts and metadata  
**Does:** keeps bacterial genera, matches samples, removes duplicate subject/time observations, and keeps subjects with at least 3 samples  
**Output:** one clean count table, metadata table, and genus taxonomy table

Countries stay separate throughout the pipeline: **FIN, EST, RUS**.

```r
COUNTS_FILE <- "../data/DIABIMMUNE/diabimmune_data_16s.csv"
METADATA_FILE <- "../data/DIABIMMUNE/diabimmune_metadata.csv"

MIN_SAMPLES_PER_SUBJECT <- 3
```

```r
root <- if (dir.exists("data")) "." else ".."  # project directory

# local copies
counts_raw <- read.csv(COUNTS_FILE, check.names = FALSE)
metadata <- read.csv(METADATA_FILE, check.names = FALSE)

# prune metadata
metadata <- data.frame(
  sample_id = as.character(metadata$SampleID),
  subject_id = as.character(metadata$subjectID),
  age = as.numeric(metadata$age_at_collection),
  country = toupper(as.character(metadata$country)),
  stringsAsFactors = FALSE
)

# standardize country
metadata$country[metadata$country == "FINLAND"] <- "FIN"
metadata$country[metadata$country == "ESTONIA"] <- "EST"
metadata$country[metadata$country == "RUSSIA"] <- "RUS"

rownames(counts_raw) <- as.character(counts_raw[[1]])
counts <- counts_raw[-1]
counts[] <- lapply(counts, as.numeric)

# prune 16s data
archaea <- grepl("^k__Archaea(?:\\||$)", colnames(counts))
genus <- grepl("\\|g__[^|]*$", colnames(counts)) & !archaea
counts <- counts[, genus, drop = FALSE]

# standardize names and serialize unknowns
original_names <- colnames(counts)
unknown <- grepl("\\|g__$", original_names)
new_names <- original_names
new_names[unknown] <- paste0(
  sub("\\|g__$", "", original_names[unknown]),
  "|g__Unknown_", sprintf("%03d", seq_len(sum(unknown)))
)
colnames(counts) <- new_names

# separate taxonomic levels
rank_value <- function(x, prefix) {
  parts <- strsplit(x, "\\|")[[1]]
  hit <- parts[startsWith(parts, prefix)]
  if (length(hit)) hit[1] else NA_character_
}

# apply taxonomic levels in a data frame
taxonomy <- data.frame(
  feature = new_names,
  kingdom = sapply(original_names, rank_value, prefix = "k__"),
  phylum = sapply(original_names, rank_value, prefix = "p__"),
  class = sapply(original_names, rank_value, prefix = "c__"),
  order = sapply(original_names, rank_value, prefix = "o__"),
  family = sapply(original_names, rank_value, prefix = "f__"),
  genus = sapply(new_names, rank_value, prefix = "g__"),
  stringsAsFactors = FALSE
)
```

```r

metadata <- metadata[
  complete.cases(metadata) &
  metadata$country %in% c("FIN", "EST", "RUS") &
  metadata$sample_id %in% rownames(counts),
]

counts <- counts[metadata$sample_id, , drop = FALSE]
metadata$library_size <- rowSums(counts)

keep <- metadata$library_size > 0
metadata <- metadata[keep, ]
counts <- counts[keep, , drop = FALSE]

# If the same subject/time appears more than once, keep the sample with the largest library.
order_rows <- order(metadata$subject_id, metadata$age, -metadata$library_size)
metadata <- metadata[order_rows, ]
counts <- counts[order_rows, , drop = FALSE]

keep <- !duplicated(metadata[c("subject_id", "age")])
metadata <- metadata[keep, ]
counts <- counts[keep, , drop = FALSE]

visits <- table(metadata$subject_id)
keep <- metadata$subject_id %in% names(visits[visits >= MIN_SAMPLES_PER_SUBJECT])
metadata <- metadata[keep, ]
counts <- counts[keep, , drop = FALSE]

order_rows <- order(metadata$subject_id, metadata$age)
metadata <- metadata[order_rows, ]
counts <- counts[order_rows, colSums(counts) > 0, drop = FALSE]
rownames(counts) <- metadata$sample_id
row.names(metadata) <- NULL

stopifnot(!anyDuplicated(metadata$sample_id))
stopifnot(all(table(metadata$subject_id) >= MIN_SAMPLES_PER_SUBJECT))
stopifnot(setequal(unique(metadata$country), c("FIN", "EST", "RUS")))
```

```r
output <- file.path(root, "data", "processed_16s", format(Sys.time(), "%Y%m%d_%H%M%S"))
dir.create(output, recursive = TRUE)

write.csv(
  data.frame(sample_id = rownames(counts), counts, check.names = FALSE),
  file.path(output, "counts.csv"),
  row.names = FALSE
)
write.csv(metadata, file.path(output, "metadata.csv"), row.names = FALSE)
write.csv(taxonomy[taxonomy$feature %in% colnames(counts), ],
          file.path(output, "taxonomy.csv"), row.names = FALSE)

cat("Saved:", output, "\n")
cat(nrow(metadata), "samples,", length(unique(metadata$subject_id)), "subjects,",
    ncol(counts), "bacterial genera\n")
print(table(metadata$country))
```
