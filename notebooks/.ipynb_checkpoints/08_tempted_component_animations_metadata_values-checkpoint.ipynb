# 08 — TEMPTED component animations

This notebook animates each TEMPTED component over time.

* **Subjects:** $A_{ir}\phi_r(t)$
* **Features:** $B_{jr}\phi_r(t)$
* **Time:** $\phi_r(t)$

Component signs are oriented so $\phi_r(t)$ is positive at its largest absolute value.

Subjects are grouped by country. Metadata strips show standardized values for the three strongest continuous metadata associations:

* purple = high
* white = average or missing
* orange = low

The Spearman association for each metadata variable is printed below the plot.

Known taxa use genus names. Unknown genera use the closest known higher taxonomic level with its rank, such as `(F)` or `(O)`.

The feature panel shows the strongest positive and negative loadings. The lower panel tracks the temporal loading used for each animation frame.


```python
from datetime import datetime
from pathlib import Path
import shutil

import matplotlib.pyplot as plt
from matplotlib import colors as mcolors
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Patch
import numpy as np
import pandas as pd
from IPython.display import Image, Video, display

# animation settings
N_FEATURES_EACH_SIDE = 10
ASSOCIATION_TOP_N = 3
METADATA_Z_CLIP = 2.0
FPS = 12
DPI = 140
FIGSIZE = (17, 10)

root = Path(".") if Path("data").exists() else Path("..")
t_dir = sorted((root / "data" / "tempted").iterdir())[-1]
meta_dir = sorted((root / "data" / "metadata_analysis").iterdir())[-1]

output = root / "data" / "tempted_animations" / datetime.now().strftime("%Y%m%d_%H%M%S")
output.mkdir(parents=True)

# notebook 05 already saved TEMPTED scores together with country and subject metadata
subjects_with_metadata = pd.read_csv(
    meta_dir / "tempted_with_metadata.csv",
    dtype={"subject_id": str},
)
metadata_associations = pd.read_csv(meta_dir / "metadata_associations.csv")

feature_loadings = pd.read_csv(t_dir / "feature_loadings.csv")
temporal_loadings = pd.read_csv(t_dir / "temporal_loadings.csv")

components = [c for c in temporal_loadings.columns if c.startswith("component_")]

countries = ["FIN", "EST", "RUS"]
colors = {
    "FIN": "green",
    "EST": "blue",
    "RUS": "red",
}

print("TEMPTED run:", t_dir)
print("Metadata analysis run:", meta_dir)
print("Components:", components)
print("Animation output:", output)
```

```python
# helper functions

def symmetric_limit(values, pad=1.08):
    """Give a zero-centered axis a little space beyond its largest value."""
    return np.max(np.abs(values)) * pad


def component_orientation(component):
    """Make phi positive where its absolute value is largest."""
    phi = temporal_loadings[component].to_numpy(float)
    peak = np.argmax(np.abs(phi))
    return 1.0 if phi[peak] > 0 else -1.0


def component_associations(component, orientation):
    """Return the strongest TEMPTED metadata associations in the plot orientation."""
    table = metadata_associations[
        (metadata_associations["method"] == "TEMPTED")
        & (metadata_associations["dimension"] == component)
    ].copy()

    table["association"] = table["association"] * orientation
    table["absolute_association"] = table["association"].abs()

    return (
        table
        .sort_values("absolute_association", ascending=False)
        .head(ASSOCIATION_TOP_N)
        .reset_index(drop=True)
    )
```

```python
# build and save one synchronized animation for every component

outputs = []

for component in components:
    time = temporal_loadings["time"].to_numpy(float)

    # Give the component one reproducible sign orientation.
    orientation = component_orientation(component)
    phi = temporal_loadings[component].to_numpy(float) * orientation

    # Use the three strongest metadata associations for this component.
    associations = component_associations(component, orientation)
    metadata_names = associations["metadata"].tolist()

    subjects = subjects_with_metadata[
        ["subject_id", "country", component] + metadata_names
    ].copy()
    subjects[component] = subjects[component] * orientation

    # Group subjects by country and sort them by component loading.
    subjects = pd.concat(
        [
            subjects[subjects["country"] == country].sort_values(component)
            for country in countries
        ],
        ignore_index=True,
    )

    # Standardized metadata values for the static strips.
    metadata_matrix = []

    for row in associations.itertuples():
        values = subjects[row.metadata]
        z = (values - values.mean()) / values.std(ddof=0)
        metadata_matrix.append(
            np.clip(z / METADATA_Z_CLIP, -1, 1)
        )

    metadata_matrix = np.column_stack(metadata_matrix)

    # keep the strongest negative and positive feature loadings.
    features = pd.concat([
        feature_loadings.nsmallest(N_FEATURES_EACH_SIDE, component),
        feature_loadings.nlargest(N_FEATURES_EACH_SIDE, component),
    ])
    features = (
        features
        .drop_duplicates()
        .sort_values(component)
        .reset_index(drop=True)
    )

    # Unknown genera use closest named taxonomic level
    feature_labels = []

    for feature in features["feature_id"]:
        parts = str(feature).split("|")
        genus = parts[-1].replace("g__", "")

        if not genus.startswith("Unknown_"):
            feature_labels.append(genus)
            continue

        label = genus
        for prefix, suffix in [
            ("f__", "(F)"),
            ("o__", "(O)"),
            ("c__", "(C)"),
            ("p__", "(P)"),
            ("k__", "(K)"),
        ]:
            name = next(
                (part.replace(prefix, "") for part in parts if part.startswith(prefix)),
                "",
            )
            if name:
                label = f"{name} {suffix}"
                break

        feature_labels.append(label)

    subject_base = subjects[component].to_numpy(float)
    feature_base = features[component].to_numpy(float)
    subject_colors = [colors[country] for country in subjects["country"]]

    phi_max = np.max(np.abs(phi))
    subject_limit = symmetric_limit(subject_base * phi_max)
    feature_limit = symmetric_limit(feature_base * phi_max)
    time_limit = symmetric_limit(phi)

    # figure layout:
    #   top-left  = metadata strips + animated subject bars
    #   top-right = animated feature bars
    #   bottom    = temporal loading
    fig = plt.figure(figsize=FIGSIZE)

    outer = fig.add_gridspec(
        2, 2,
        width_ratios=[1.85, .70],
        height_ratios=[4.8, 1.20],
        left=0.07,
        right=0.97,
        bottom=0.12,
        top=0.90,
        wspace=0.26,
        hspace=0.30,
    )

    metadata_width = 0.075 * len(associations)

    subject_grid = outer[0, 0].subgridspec(
        1, 2,
        width_ratios=[metadata_width, 1.00],
        wspace=0.01,
    )

    ax_strip = fig.add_subplot(subject_grid[0, 0])
    ax_subject = fig.add_subplot(subject_grid[0, 1], sharey=ax_strip)
    ax_feature = fig.add_subplot(outer[0, 1])
    ax_time = fig.add_subplot(outer[1, :])

    subject_y = np.arange(len(subjects))
    feature_y = np.arange(len(features))

    # metadata strips.
    cmap = plt.get_cmap("PuOr").copy()
    cmap.set_bad("white")

    ax_strip.imshow(
        np.ma.masked_invalid(metadata_matrix),
        aspect="auto",
        origin="lower",
        interpolation="nearest",
        cmap=cmap,
        norm=mcolors.TwoSlopeNorm(vmin=-1, vcenter=0, vmax=1),
        extent=(-0.5, len(associations) - 0.5, -0.5, len(subjects) - 0.5),
    )

    ax_strip.set_xlim(-0.5, len(associations) - 0.5)
    ax_strip.set_ylim(-0.5, len(subjects) - 0.5)
    ax_strip.set_xticks(np.arange(len(associations)))

    metadata_labels = []
    for name in metadata_names:
        label = str(name).replace("_", " ")
        metadata_labels.append(label if len(label) <= 20 else label[:19] + "…")

    ax_strip.set_xticklabels(
        metadata_labels,
        rotation=30,
        ha="left",
        va="bottom",
        fontsize=14,
    )
    ax_strip.tick_params(
        axis="x",
        top=True,
        labeltop=True,
        bottom=False,
        labelbottom=False,
        pad=2,
    )
    

    # Animated subject and feature bars.
    subject_bars = ax_subject.barh(
        subject_y,
        subject_base * phi[0],
        color=subject_colors,
        alpha=0.82,
    )

    feature_bars = ax_feature.barh(
        feature_y,
        feature_base * phi[0],
        color="0.45",
        alpha=0.85,
    )

    ax_subject.axvline(0, color="black", linewidth=1)
    ax_subject.set_xlim(-subject_limit, subject_limit)
    ax_subject.set_yticks([])
    ax_subject.set_xlabel(r"Time-scaled subject contribution  $A_i\phi(t)$")
    ax_subject.set_title("Subjects")
    ax_subject.grid(axis="x", alpha=0.15)

    # Country dividers and labels.
    country_groups = []
    for country in countries:
        positions = np.where(subjects["country"].to_numpy() == country)[0]
        country_groups.append((country, positions[0], positions[-1]))

    for left_group, right_group in zip(country_groups[:-1], country_groups[1:]):
        boundary = (left_group[2] + right_group[1]) / 2
        ax_strip.axhline(boundary, color="black", linewidth=1.2)
        ax_subject.axhline(boundary, color="black", linewidth=1.2)

    label_x = -subject_limit * 0.98

    for country, start, end in country_groups:
        ax_subject.text(
            label_x,
            (start + end) / 2,
            country,
            ha="left",
            va="center",
            fontsize=12,
            fontweight="bold",
            bbox=dict(facecolor="white", edgecolor="none", alpha=0.78, pad=1.5),
        )

    ax_feature.axvline(0, color="black", linewidth=1)
    ax_feature.set_xlim(-feature_limit, feature_limit)
    ax_feature.set_yticks(feature_y)
    ax_feature.set_yticklabels(feature_labels, fontsize=12)
    ax_feature.set_xlabel(r"Time-scaled feature contribution  $B_j\phi(t)$")
    ax_feature.set_title("Features")
    ax_feature.grid(axis="x", alpha=0.15)

    # Temporal loading.
    ax_time.plot(time, phi, color="black", linewidth=1.5)
    ax_time.axhline(0, color="black", linewidth=0.7, alpha=0.6)
    marker, = ax_time.plot([time[0]], [phi[0]], "o", color="black", markersize=6)

    ax_time.set_xlim(time.min(), time.max())
    ax_time.set_ylim(-time_limit, time_limit)
    ax_time.set_xlabel("Age")
    ax_time.set_ylabel(r"$\phi(t)$")
    ax_time.set_title(f"{component}: oriented temporal loading", fontsize=14)
    ax_time.grid(alpha=0.15)

    # Country legend.
    country_handles = [
        plt.Line2D(
            [0], [0],
            marker="s",
            linestyle="",
            markersize=7,
            markerfacecolor=colors[country],
            markeredgecolor=colors[country],
            label=country,
        )
        for country in countries
    ]

    country_legend = ax_subject.legend(
        handles=country_handles,
        title="Country",
        loc="lower right",
        fontsize=12,
        title_fontsize=14,
    )
    ax_subject.add_artist(country_legend)

    # Metadata-value legend.
    strip_handles = [
        Patch(facecolor=cmap(0.85), label="High"),
        Patch(facecolor="white", edgecolor="0.7", label="Average / missing"),
        Patch(facecolor=cmap(0.15), label="Low"),
    ]

    ax_subject.legend(
        handles=strip_handles,
        title="Metadata values",
        loc="upper right",
        fontsize=12,
        title_fontsize=14,
    )

    association_text = " | ".join(
        f"{row.metadata} ({row.association:+.2f})"
        for row in associations.itertuples()
    )

    fig.text(
        0.07,
        0.045,
        f"Metadata associations with oriented subject loading: {association_text}",
        ha="left",
        va="center",
        fontsize=12,
    )

    frame_label = fig.text(
        0.5,
        0.965,
        "",
        ha="center",
        va="top",
        fontsize=12,
    )

    # FuncAnimation repeatedly calls this function for each time point.
    def update(frame):
        p = phi[frame]

        for bar, value in zip(subject_bars, subject_base * p):
            bar.set_width(value)

        for bar, value in zip(feature_bars, feature_base * p):
            bar.set_width(value)

        marker.set_data([time[frame]], [p])
        frame_label.set_text(
            f"{component} | age = {time[frame]:.1f} | temporal loading = {p:.3f}"
        )

        return [*subject_bars, *feature_bars, marker, frame_label]

    animation = FuncAnimation(
        fig,
        update,
        frames=len(time),
        interval=1000 / FPS,
        blit=False,
        repeat=True,
    )

    if shutil.which("ffmpeg"):
        animation_path = output / f"{component}_tempted_contributions.mp4"
        animation.save(animation_path, writer="ffmpeg", fps=FPS, dpi=DPI)
    else:
        animation_path = output / f"{component}_tempted_contributions.gif"
        animation.save(animation_path, writer=PillowWriter(fps=FPS), dpi=DPI)

    plt.close(fig)

    outputs.append({
        "component": component,
        "animation": str(animation_path),
    })
    print("Saved:", animation_path)

pd.DataFrame(outputs)
```

```python
# show the metadata associations used in each animation

association_summaries = []

for component in components:
    orientation = component_orientation(component)
    table = component_associations(component, orientation).copy()
    table.insert(0, "component", component)
    table.insert(1, "orientation", int(orientation))
    association_summaries.append(table)

association_summary = pd.concat(association_summaries, ignore_index=True)

display(
    association_summary[
        ["component", "orientation", "metadata", "association", "absolute_association"]
    ]
)
```

```python
# preview every generated animation

for row in outputs:
    component = row["component"]
    path = Path(row["animation"])

    print(component)

    if path.suffix.lower() == ".mp4":
        display(Video(str(path), embed=True, html_attributes="controls loop"))
    else:
        display(Image(filename=str(path)))
```
