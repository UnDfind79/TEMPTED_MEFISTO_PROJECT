# 05 — Post-method train/test evaluation

**Input:** full-cohort TEMPTED and MEFISTO subject scores  
**Does:** creates 20 repeated 70/30 subject-level splits *after* both methods have been fit, then trains the same country classifier on each method's latent dimensions  
**Output:** split assignments, predictions, metrics, and per-country recall

This notebook probes how much country information is present in the already learned latent representations.

```python
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

N_SPLITS = 20
TRAIN_SIZE = 0.70
SEED = 2026
COUNTRIES = ["FIN", "EST", "RUS"]

root = Path(".") if Path("data").exists() else Path("..")
tempted_folder = sorted((root / "data" / "tempted").iterdir())[-1]
mefisto_folder = sorted((root / "data" / "mefisto").iterdir())[-1]
run_id = datetime.now().strftime("%Y%m%d_%H%M%S")
output = root / "data" / "evaluation" / run_id
output.mkdir(parents=True, exist_ok=True)

tempted = pd.read_csv(tempted_folder / "subject_scores.csv", dtype={"subject_id": str, "country": str})
mefisto = pd.read_csv(mefisto_folder / "subject_scores.csv", dtype={"subject_id": str, "country": str})

tempted_columns = [c for c in tempted.columns if c.startswith("component_")]
mefisto_columns = [c for c in mefisto.columns if c.startswith("factor_")]

tempted = tempted[["subject_id", "country"] + tempted_columns].copy()
mefisto = mefisto[["subject_id", "country"] + mefisto_columns].copy()

tempted = tempted[tempted["country"].isin(COUNTRIES)].copy()
mefisto = mefisto[mefisto["country"].isin(COUNTRIES)].copy()

subjects = tempted[["subject_id", "country"]].merge(
    mefisto[["subject_id", "country"]], on=["subject_id", "country"], how="inner"
).drop_duplicates().sort_values("subject_id")

if subjects["country"].nunique() != 3:
    raise ValueError("Expected exactly FIN, EST, and RUS in the subject table.")

print("TEMPTED:", tempted_folder, "dimensions:", len(tempted_columns))
print("MEFISTO:", mefisto_folder, "dimensions:", len(mefisto_columns))
print("Output:", output)
```

## Create the splits and evaluate both methods

Each split is generated once and reused for TEMPTED and MEFISTO. Standardization is learned from training subjects only.

```python
split_rows = []
prediction_rows = []
metric_rows = []
country_rows = []

for split in range(1, N_SPLITS + 1):
    train_subjects, test_subjects = train_test_split(
        subjects,
        train_size=TRAIN_SIZE,
        stratify=subjects["country"],
        random_state=SEED + split,
    )

    for frame, which in [(train_subjects, "train"), (test_subjects, "test")]:
        temp = frame.copy()
        temp["split"] = split
        temp["set"] = which
        split_rows.append(temp)

    for method, scores, columns in [
        ("TEMPTED", tempted, tempted_columns),
        ("MEFISTO", mefisto, mefisto_columns),
    ]:
        train = scores[scores["subject_id"].isin(train_subjects["subject_id"])].copy()
        test = scores[scores["subject_id"].isin(test_subjects["subject_id"])].copy()

        scaler = StandardScaler()
        x_train = scaler.fit_transform(train[columns])
        x_test = scaler.transform(test[columns])

        model = LogisticRegression(max_iter=2000, multi_class="multinomial")
        model.fit(x_train, train["country"])
        predicted = model.predict(x_test)

        metric_rows.append({
            "split": split,
            "method": method,
            "n_dimensions": len(columns),
            "accuracy": accuracy_score(test["country"], predicted),
            "balanced_accuracy": balanced_accuracy_score(test["country"], predicted),
            "macro_f1": f1_score(test["country"], predicted, average="macro"),
        })

        recalls = recall_score(test["country"], predicted, labels=COUNTRIES, average=None)
        for country, value in zip(COUNTRIES, recalls):
            country_rows.append({"split": split, "method": method, "country": country, "recall": value})

        out = test[["subject_id", "country"]].copy()
        out["split"] = split
        out["method"] = method
        out["truth"] = out["country"]
        out["predicted"] = predicted
        prediction_rows.append(out[["split", "method", "subject_id", "truth", "predicted"]])
```

```python
splits = pd.concat(split_rows, ignore_index=True)
predictions = pd.concat(prediction_rows, ignore_index=True)
metrics = pd.DataFrame(metric_rows)
country_metrics = pd.DataFrame(country_rows)

for split in range(1, N_SPLITS + 1):
    test_countries = splits[(splits["split"] == split) & (splits["set"] == "test")]["country"]
    if sorted(test_countries.unique()) != sorted(COUNTRIES):
        raise ValueError(f"Split {split} test set does not contain all three countries.")

metric_summary = metrics.groupby("method")[["accuracy", "balanced_accuracy", "macro_f1"]].agg(["mean", "std", "median"])
dimension_summary = pd.DataFrame({
    "method": ["TEMPTED", "MEFISTO"],
    "n_dimensions": [len(tempted_columns), len(mefisto_columns)],
})

splits.to_csv(output / "split_assignments.csv", index=False)
predictions.to_csv(output / "predictions.csv", index=False)
metrics.to_csv(output / "metrics.csv", index=False)
metric_summary.to_csv(output / "metric_summary.csv")
country_metrics.to_csv(output / "per_country_recall.csv", index=False)
dimension_summary.to_csv(output / "dimension_summary.csv", index=False)

print("Saved:", output)
metric_summary
```
