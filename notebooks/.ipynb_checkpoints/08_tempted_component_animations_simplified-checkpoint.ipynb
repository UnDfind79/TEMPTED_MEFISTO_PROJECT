# 08 — TEMPTED component animations

This notebook animates the subject, feature, and temporal parts of each TEMPTED component.

For component \(r\), the animated values are:

- **Subject contribution:** \(A_{ir}\phi_r(t)\)
- **Feature contribution:** \(B_{jr}\phi_r(t)\)

The subject and feature bars are centered at zero. As the temporal loading \(\phi_r(t)\) changes, the bars grow, shrink, or reverse direction. The lower panel shows the same temporal loading and moves a point along the curve.

## Component orientation

TEMPTED component signs are arbitrary, so each component is given one consistent orientation. Find the time where \(|\phi_r(t)|\) is largest. If \(\phi_r(t)\) is negative there, multiply both the temporal loading and subject loading by \(-1\).

This leaves the fitted component unchanged but gives every plot a reproducible direction. Metadata-association signs use the same orientation.

## Subject panel

Every TEMPTED subject is shown. Subjects are grouped as **FIN, EST, RUS** and sorted by their oriented subject loading within each country.

- **Bar color:** country
- **Metadata strips:** the three strongest continuous subject-level metadata associations from notebook 05
- **Animated bars:** \(A_{ir}\phi_r(t)\)

The metadata strips are static because the metadata values and subject loadings are static. Purple means the subject's metadata value agrees with the direction expected from the component association, orange means it points in the opposite direction, and white is near neutral or missing.

## Feature labels

Notebook 01 keeps genus-level bacterial features and names an empty genus as `Unknown_###`.

Known genera are displayed by genus name alone. For an `Unknown_###` genus, the label uses the closest named higher taxonomic level and adds its rank:

- `(F)` family
- `(O)` order
- `(C)` class
- `(P)` phylum
- `(K)` kingdom

For example, an unknown genus within `Lachnospiraceae` is displayed as `Lachnospiraceae (F)`.

## Feature and temporal panels

The feature panel shows the strongest negative and positive feature loadings multiplied by the same oriented temporal loading. The temporal panel shows \(\phi_r(t)\) directly.

If the animated bars reverse direction, the temporal loading has changed sign. The subject metadata associations themselves do not change over time.


```python
from datetime import datetime
from pathlib import Path
import shutil

import matplotlib.pyplot as plt
from matplotlib import colors as mcolors
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Patch
import numpy as np
import pandas as pd
from IPython.display import Image, Video, display

# animation settings
N_FEATURES_EACH_SIDE = 10
ASSOCIATION_TOP_N = 3
METADATA_Z_CLIP = 2.0
FPS = 12
DPI = 140
FIGSIZE = (17, 10)

root = Path(".") if Path("data").exists() else Path("..")
t_dir = sorted((root / "data" / "tempted").iterdir())[-1]
meta_dir = sorted((root / "data" / "metadata_analysis").iterdir())[-1]

output = root / "data" / "tempted_animations" / datetime.now().strftime("%Y%m%d_%H%M%S")
output.mkdir(parents=True)

# notebook 05 already saved TEMPTED scores together with country and subject metadata
subjects_with_metadata = pd.read_csv(
    meta_dir / "tempted_with_metadata.csv",
    dtype={"subject_id": str},
)
metadata_associations = pd.read_csv(meta_dir / "metadata_associations.csv")

feature_loadings = pd.read_csv(t_dir / "feature_loadings.csv")
temporal_loadings = pd.read_csv(t_dir / "temporal_loadings.csv")

components = [c for c in temporal_loadings.columns if c.startswith("component_")]

countries = ["FIN", "EST", "RUS"]
colors = {
    "FIN": "green",
    "EST": "blue",
    "RUS": "red",
}

print("TEMPTED run:", t_dir)
print("Metadata analysis run:", meta_dir)
print("Components:", components)
print("Animation output:", output)
```

```python
# helper functions

def short_feature(feature):
    """Show known genera normally; replace Unknown_### with the nearest named higher rank."""
    parts = str(feature).split("|")
    genus = parts[-1].replace("g__", "")

    if not genus.startswith("Unknown_"):
        return genus

    for prefix, suffix in [
        ("f__", "(F)"),
        ("o__", "(O)"),
        ("c__", "(C)"),
        ("p__", "(P)"),
        ("k__", "(K)"),
    ]:
        name = next(
            (part.replace(prefix, "") for part in parts if part.startswith(prefix)),
            "",
        )
        if name:
            return f"{name} {suffix}"

    return genus


def short_metadata(name, max_len=20):
    """Make metadata names compact enough for the strip headers."""
    name = str(name).replace("_", " ")
    return name if len(name) <= max_len else name[:max_len - 1] + "…"


def strongest_both_sides(frame, column, n):
    """Keep the strongest negative and positive feature loadings."""
    return (
        pd.concat([
            frame.nsmallest(n, column),
            frame.nlargest(n, column),
        ])
        .drop_duplicates()
        .sort_values(column)
        .reset_index(drop=True)
    )


def symmetric_limit(values, pad=1.08):
    """Give a zero-centered axis a little space beyond its largest value."""
    return np.max(np.abs(values)) * pad


def component_orientation(component):
    """Make phi positive where its absolute value is largest."""
    phi = temporal_loadings[component].to_numpy(float)
    peak = np.argmax(np.abs(phi))
    return 1.0 if phi[peak] > 0 else -1.0


def component_associations(component, orientation):
    """Return the strongest TEMPTED metadata associations in the plot orientation."""
    table = metadata_associations[
        (metadata_associations["method"] == "TEMPTED")
        & (metadata_associations["dimension"] == component)
    ].copy()

    table["association"] = table["association"] * orientation
    table["absolute_association"] = table["association"].abs()

    return (
        table
        .sort_values("absolute_association", ascending=False)
        .head(ASSOCIATION_TOP_N)
        .reset_index(drop=True)
    )


def metadata_concordance(subjects, component, associations):
    """
    Build the static metadata strips.

    concordance = sign(subject loading) * sign(metadata association) * z(metadata)
    """
    matrix = []
    subject_sign = np.sign(subjects[component].to_numpy(float))

    for row in associations.itertuples():
        values = subjects[row.metadata]
        z = (values - values.mean()) / values.std(ddof=0)

        concordance = subject_sign * np.sign(row.association) * z
        concordance = np.clip(concordance / METADATA_Z_CLIP, -1, 1)
        matrix.append(concordance)

    return np.column_stack(matrix)
```

```python
# build one synchronized animation for one component

def animate_component(component):
    time = temporal_loadings["time"].to_numpy(float)

    # Give the component one reproducible sign orientation.
    orientation = component_orientation(component)
    phi = temporal_loadings[component].to_numpy(float) * orientation

    # Pick the metadata columns first, then keep them beside each subject.
    associations = component_associations(component, orientation)
    metadata_names = associations["metadata"].tolist()

    subjects = subjects_with_metadata[
        ["subject_id", "country", component] + metadata_names
    ].copy()
    subjects[component] = subjects[component] * orientation

    # Group subjects by country and sort them by their component loading.
    subjects = pd.concat(
        [
            subjects[subjects["country"] == country].sort_values(component)
            for country in countries
        ],
        ignore_index=True,
    )

    metadata_matrix = metadata_concordance(
        subjects,
        component,
        associations,
    )

    # Keep strong feature loadings from both directions.
    features = strongest_both_sides(
        feature_loadings[["feature_id", component]],
        component,
        N_FEATURES_EACH_SIDE,
    )
    features["label"] = features["feature_id"].map(short_feature)

    subject_base = subjects[component].to_numpy(float)
    feature_base = features[component].to_numpy(float)

    subject_colors = [colors[country] for country in subjects["country"]]
    feature_labels = features["label"].tolist()

    phi_max = np.max(np.abs(phi))
    subject_limit = symmetric_limit(subject_base * phi_max)
    feature_limit = symmetric_limit(feature_base * phi_max)
    time_limit = symmetric_limit(phi)

    # Figure layout:
    #   top-left  = metadata strips + animated subject bars
    #   top-right = animated feature bars
    #   bottom    = temporal loading
    fig = plt.figure(figsize=FIGSIZE)

    outer = fig.add_gridspec(
        2, 2,
        width_ratios=[1.85, 1.00],
        height_ratios=[4.8, 1.20],
        left=0.07,
        right=0.97,
        bottom=0.12,
        top=0.90,
        wspace=0.26,
        hspace=0.30,
    )

    metadata_width = 0.075 * len(associations)

    subject_grid = outer[0, 0].subgridspec(
        1, 2,
        width_ratios=[metadata_width, 1.00],
        wspace=0.01,
    )

    ax_strip = fig.add_subplot(subject_grid[0, 0])
    ax_subject = fig.add_subplot(subject_grid[0, 1], sharey=ax_strip)
    ax_feature = fig.add_subplot(outer[0, 1])
    ax_time = fig.add_subplot(outer[1, :])

    subject_y = np.arange(len(subjects))
    feature_y = np.arange(len(features))

    # Static metadata strips.
    cmap = plt.get_cmap("PuOr").copy()
    cmap.set_bad("white")

    ax_strip.imshow(
        np.ma.masked_invalid(metadata_matrix),
        aspect="auto",
        origin="lower",
        interpolation="nearest",
        cmap=cmap,
        norm=mcolors.TwoSlopeNorm(vmin=-1, vcenter=0, vmax=1),
        extent=(-0.5, len(associations) - 0.5, -0.5, len(subjects) - 0.5),
    )

    ax_strip.set_xlim(-0.5, len(associations) - 0.5)
    ax_strip.set_ylim(-0.5, len(subjects) - 0.5)
    ax_strip.set_xticks(np.arange(len(associations)))
    ax_strip.set_xticklabels(
        [short_metadata(name) for name in metadata_names],
        rotation=60,
        ha="left",
        va="bottom",
        fontsize=7,
    )
    ax_strip.tick_params(
        axis="x",
        top=True,
        labeltop=True,
        bottom=False,
        labelbottom=False,
        pad=2,
    )
    ax_strip.set_yticks([])
    ax_strip.set_ylabel(
        "SUBJECTS",
        rotation=90,
        fontsize=11,
        fontweight="bold",
        labelpad=18,
    )
    ax_strip.set_title("Metadata concordance", fontsize=9, pad=8)

    # Animated subject and feature bars.
    subject_bars = ax_subject.barh(
        subject_y,
        subject_base * phi[0],
        color=subject_colors,
        alpha=0.82,
    )

    feature_bars = ax_feature.barh(
        feature_y,
        feature_base * phi[0],
        color="0.45",
        alpha=0.85,
    )

    ax_subject.axvline(0, color="black", linewidth=1)
    ax_subject.set_xlim(-subject_limit, subject_limit)
    ax_subject.set_yticks([])
    ax_subject.set_xlabel(r"Time-scaled subject contribution  $A_i\phi(t)$")
    ax_subject.set_title("Subjects")
    ax_subject.grid(axis="x", alpha=0.15)

    # Country dividers and labels.
    country_groups = []
    for country in countries:
        positions = np.where(subjects["country"].to_numpy() == country)[0]
        country_groups.append((country, positions[0], positions[-1]))

    for left_group, right_group in zip(country_groups[:-1], country_groups[1:]):
        boundary = (left_group[2] + right_group[1]) / 2
        ax_strip.axhline(boundary, color="black", linewidth=1.2)
        ax_subject.axhline(boundary, color="black", linewidth=1.2)

    label_x = -subject_limit * 0.98

    for country, start, end in country_groups:
        ax_subject.text(
            label_x,
            (start + end) / 2,
            country,
            ha="left",
            va="center",
            fontsize=8,
            fontweight="bold",
            bbox=dict(facecolor="white", edgecolor="none", alpha=0.78, pad=1.5),
        )

    ax_feature.axvline(0, color="black", linewidth=1)
    ax_feature.set_xlim(-feature_limit, feature_limit)
    ax_feature.set_yticks(feature_y)
    ax_feature.set_yticklabels(feature_labels, fontsize=8)
    ax_feature.set_xlabel(r"Time-scaled feature contribution  $B_j\phi(t)$")
    ax_feature.set_title("Features")
    ax_feature.grid(axis="x", alpha=0.15)

    # Temporal loading.
    ax_time.plot(time, phi, color="black", linewidth=1.5)
    ax_time.axhline(0, color="black", linewidth=0.7, alpha=0.6)
    marker, = ax_time.plot([time[0]], [phi[0]], "o", color="black", markersize=6)

    ax_time.set_xlim(time.min(), time.max())
    ax_time.set_ylim(-time_limit, time_limit)
    ax_time.set_xlabel("Age")
    ax_time.set_ylabel(r"$\phi(t)$")
    ax_time.set_title(f"{component}: oriented temporal loading", fontsize=10)
    ax_time.grid(alpha=0.15)

    # Country legend.
    country_handles = [
        plt.Line2D(
            [0], [0],
            marker="s",
            linestyle="",
            markersize=7,
            markerfacecolor=colors[country],
            markeredgecolor=colors[country],
            label=country,
        )
        for country in countries
    ]

    country_legend = ax_subject.legend(
        handles=country_handles,
        title="Country",
        loc="lower right",
        fontsize=8,
        title_fontsize=8,
    )
    ax_subject.add_artist(country_legend)

    # Metadata-strip legend.
    strip_handles = [
        Patch(facecolor=cmap(0.85), label="Concordant"),
        Patch(facecolor="white", edgecolor="0.7", label="Weak / neutral"),
        Patch(facecolor=cmap(0.15), label="Discordant"),
    ]

    ax_subject.legend(
        handles=strip_handles,
        title="Metadata strips",
        loc="upper right",
        fontsize=8,
        title_fontsize=8,
    )

    association_text = " | ".join(
        f"{row.metadata} ({row.association:+.2f})"
        for row in associations.itertuples()
    )

    fig.text(
        0.07,
        0.045,
        f"Metadata associations with oriented subject loading: {association_text}",
        ha="left",
        va="center",
        fontsize=8,
    )

    frame_label = fig.text(
        0.5,
        0.965,
        "",
        ha="center",
        va="top",
        fontsize=12,
    )

    def update(frame):
        p = phi[frame]

        for bar, value in zip(subject_bars, subject_base * p):
            bar.set_width(value)

        for bar, value in zip(feature_bars, feature_base * p):
            bar.set_width(value)

        marker.set_data([time[frame]], [p])
        frame_label.set_text(
            f"{component} | age = {time[frame]:.1f} | temporal loading = {p:.3f}"
        )

        return [*subject_bars, *feature_bars, marker, frame_label]

    animation = FuncAnimation(
        fig,
        update,
        frames=len(time),
        interval=1000 / FPS,
        blit=False,
        repeat=True,
    )

    if shutil.which("ffmpeg"):
        animation_path = output / f"{component}_tempted_contributions.mp4"
        animation.save(animation_path, writer="ffmpeg", fps=FPS, dpi=DPI)
    else:
        animation_path = output / f"{component}_tempted_contributions.gif"
        animation.save(animation_path, writer=PillowWriter(fps=FPS), dpi=DPI)

    plt.close(fig)
    return animation_path
```

```python
# show the metadata associations used in each animation

association_summaries = []

for component in components:
    orientation = component_orientation(component)
    table = component_associations(component, orientation).copy()
    table.insert(0, "component", component)
    table.insert(1, "orientation", int(orientation))
    association_summaries.append(table)

association_summary = pd.concat(association_summaries, ignore_index=True)

display(
    association_summary[
        ["component", "orientation", "metadata", "association", "absolute_association"]
    ]
)
```

```python
# render every TEMPTED component

outputs = []

for component in components:
    animation_path = animate_component(component)
    outputs.append({
        "component": component,
        "animation": str(animation_path),
    })
    print("Saved:", animation_path)

pd.DataFrame(outputs)
```

```python
# preview every generated animation

for row in outputs:
    component = row["component"]
    path = Path(row["animation"])

    print(component)

    if path.suffix.lower() == ".mp4":
        display(Video(str(path), embed=True, html_attributes="controls loop"))
    else:
        display(Image(filename=str(path)))
```

## Interpretation

TEMPTED separates each component into subject, feature, and temporal parts. The animation multiplies the fixed subject and feature loadings by the changing temporal loading.

### Orientation

For each component, \(\phi_r(t)\) is made positive at the time where \(|\phi_r(t)|\) is largest. If a sign flip is needed, the subject loading is flipped with it. The fitted component is unchanged, and the metadata associations use the same orientation.

### Subject bars

Each subject bar shows

\[
A_{ir}\phi_r(t).
\]

When \(\phi_r(t)\) approaches zero, all subject bars approach zero. When \(\phi_r(t)\) changes sign, the bars reverse direction. Subjects stay grouped by country and ordered by their fixed \(A_{ir}\) values.

### Metadata strips

Notebook 05 calculates Spearman associations between each TEMPTED subject component and six continuous subject-level metadata variables. The three strongest associations for each component are shown.

For metadata variable \(x\) with association \(\rho\), each strip value is

\[
\operatorname{sign}(A_{ir})
\operatorname{sign}(\rho)
z(x_i).
\]

The value is divided by `METADATA_Z_CLIP` and clipped to \([-1,1]\) for color display.

- **Purple:** concordant
- **White:** weak, neutral, or missing
- **Orange:** discordant

The association coefficient for each strip is printed below the animation. Country is shown separately by subject-bar color.

### Feature bars

Each feature bar shows

\[
B_{jr}\phi_r(t).
\]

The plot keeps the strongest negative and positive loadings so both sides of the component remain visible.

Known genus names are shown directly. Features that notebook 01 serialized as `Unknown_###` use the closest named higher taxonomic level followed by its rank, such as `(F)` or `(O)`.

### Temporal loading

The lower panel shows \(\phi_r(t)\). The moving point marks the age used by the current animation frame.

`N_FEATURES_EACH_SIDE`, `ASSOCIATION_TOP_N`, `METADATA_Z_CLIP`, and `FPS` control the number of features, number of metadata strips, strip color saturation, and animation speed.
